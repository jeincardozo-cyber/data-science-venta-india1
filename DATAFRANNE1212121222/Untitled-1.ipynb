```python
# detencion e valores extremos
#OUTLIERS - MÉTODO IQR

def detectar_outliers(data, columna):

    Q1 = data[columna].quantile(0.25)
    Q3 = data[columna].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = data[
        (data[columna] < limite_inferior) |
        (data[columna] > limite_superior)
    ]

    return outliers


for columna in columnas_numericas:

    cantidad = len(
        detectar_outliers(
            df,
            columna
        )
    )

    print(
        f"{columna}: {cantidad} posibles outliers"
    )
```

```python
# Calcular Q1
Q1 = datos.quantile(0.25)
```

```python
# Calcular Q3
Q3 = datos.quantile(0.75)
```

```python
# Calcular IQR
IQR = Q3 - Q1
```

```python
# Límites para detectar posibles valores atípicos
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR
```

```python
# Cantidad de outliers por columna
outliers = ((datos < limite_inferior) | (datos > limite_superior)).sum()
```

```python
# Porcentaje de outliers
porcentaje_outliers = (outliers / datos.notna().sum()) * 100
```

```python
# Crear tabla de resultados
resultado_outliers = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Límite inferior": limite_inferior,
    "Límite superior": limite_superior,
    "Outliers": outliers,
    "% Outliers": porcentaje_outliers
})
```
