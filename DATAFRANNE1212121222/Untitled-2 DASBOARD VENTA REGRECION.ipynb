```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Usar una muestra para calcular Silhouette
np.random.seed(42)

TAMANO_MUESTRA = 10000

if len(X_cluster) > TAMANO_MUESTRA:
    indices = np.random.choice(
        len(X_cluster),
        TAMANO_MUESTRA,
        replace=False
    )
    X_muestra = X_cluster[indices]
else:
    X_muestra = X_cluster

print("Registros totales:", len(X_cluster))
print("Registros utilizados para Silhouette:", len(X_muestra))
```

```python
resultados_kmeans = []

for k in range(2, 9):
    modelo_k = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    grupos = modelo_k.fit_predict(X_muestra)

    puntuacion = silhouette_score(
        X_muestra,
        grupos
    )

    resultados_kmeans.append([
        k,
        puntuacion
    ])

tabla_kmeans = pd.DataFrame(
    resultados_kmeans,
    columns=["k", "silhouette"]
)

print("RESULTADO K-MEANS")
print("=" * 50)

print(
    tabla_kmeans.round(4).to_string(index=False)
)
```

```python
#automáticamente el mejor K


K_OPTIMO = int(
    tabla_kmeans.loc[
        tabla_kmeans["silhouette"].idxmax(),
        "k"
    ]
)

MEJOR_SILHOUETTE = tabla_kmeans.loc[
    tabla_kmeans["silhouette"].idxmax(),
    "silhouette"
]

print("K ÓPTIMO:", K_OPTIMO)
print(
    "MEJOR SILHOUETTE:",
    round(MEJOR_SILHOUETTE, 4)
)
```

```python
#K-Means sobre los 108.802 registros
modelo_kmeans = KMeans(
    n_clusters=K_OPTIMO,
    random_state=42,
    n_init=10
)

clusters = modelo_kmeans.fit_predict(
    X_cluster
)

pca_df["cluster"] = clusters

print("K-MEANS FINALIZADO")
print("=" * 50)
print("Registros procesados:", len(clusters))
print("Número de clusters:", K_OPTIMO)
```

```python
#Árbol de Decisión
modelo_arbol = DecisionTreeRegressor(
    max_depth=12,
    random_state=42
)

modelo_arbol.fit(
    X_train,
    y_train
)

pred_arbol = modelo_arbol.predict(
    X_test
)

mae_arbol = mean_absolute_error(
    y_test,
    pred_arbol
)

rmse_arbol = np.sqrt(
    mean_squared_error(
        y_test,
        pred_arbol
    )
)

r2_arbol = r2_score(
    y_test,
    pred_arbol
)

print("ÁRBOL DE DECISIÓN")
print("MAE :", round(mae_arbol, 2))
print("RMSE:", round(rmse_arbol, 2))
print("R²  :", round(r2_arbol, 4))
```

```python
#RAMDON FOREST

modelo_random_forest = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

modelo_random_forest.fit(
    X_train,
    y_train
)

pred_random_forest = modelo_random_forest.predict(
    X_test
)

mae_rf = mean_absolute_error(
    y_test,
    pred_random_forest
)

rmse_rf = np.sqrt(
    mean_squared_error(
        y_test,
        pred_random_forest
    )
)

r2_rf = r2_score(
    y_test,
    pred_random_forest
)

print("RANDOM FOREST")
print("MAE :", round(mae_rf, 2))
print("RMSE:", round(rmse_rf, 2))
print("R²  :", round(r2_rf, 4))
```

```python
# SUPPORT VECTOR REGRESSION (SVR)

escalador_svr = StandardScaler()

X_train_svr = escalador_svr.fit_transform(
    X_train
)

X_test_svr = escalador_svr.transform(
    X_test
)

modelo_svr = SVR(
    kernel="rbf",
    C=100,
    epsilon=0.1
)

modelo_svr.fit(
    X_train_svr,
    y_train
)

pred_svr = modelo_svr.predict(
    X_test_svr
)

mae_svr = mean_absolute_error(
    y_test,
    pred_svr
)

rmse_svr = np.sqrt(
    mean_squared_error(
        y_test,
        pred_svr
    )
)

r2_svr = r2_score(
    y_test,
    pred_svr
)

print("SUPPORT VECTOR REGRESSION")
print("MAE :", round(mae_svr, 2))
print("RMSE:", round(rmse_svr, 2))
print("R²  :", round(r2_svr, 4))
```

```python
#KNN REGRECCION
modelo_knn = KNeighborsRegressor(
    n_neighbors=5,
    weights="distance"
)

modelo_knn.fit(
    X_train_svr,
    y_train
)

pred_knn = modelo_knn.predict(
    X_test_svr
)

mae_knn = mean_absolute_error(
    y_test,
    pred_knn
)

rmse_knn = np.sqrt(
    mean_squared_error(
        y_test,
        pred_knn
    )
)

r2_knn = r2_score(
    y_test,
    pred_knn
)

print("KNN REGRESSION")
print("MAE :", round(mae_knn, 2))
print("RMSE:", round(rmse_knn, 2))
print("R²  :", round(r2_knn, 4))
```

```python
#Gradient Boosting
modelo_gradient = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=5,
    random_state=42
)

modelo_gradient.fit(
    X_train,
    y_train
)

pred_gradient = modelo_gradient.predict(
    X_test
)

mae_gradient = mean_absolute_error(
    y_test,
    pred_gradient
)

rmse_gradient = np.sqrt(
    mean_squared_error(
        y_test,
        pred_gradient
    )
)

r2_gradient = r2_score(
    y_test,
    pred_gradient
)

print("GRADIENT BOOSTING")
print("MAE :", round(mae_gradient, 2))
print("RMSE:", round(rmse_gradient, 2))
print("R²  :", round(r2_gradient, 4))
```

```python

# DASHBOARD ANIMADO - VENTA INDIA
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# ============================================================
# PREPARAR DATOS
# ============================================================

# Ventas mensuales
ventas_mes = (
    df_limpio
    .groupby("month")["net_sales"]
    .sum()
    .sort_index()
)

# Ganancia mensual
ganancia_mes = (
    df_limpio
    .groupby("month")["profit_amount"]
    .sum()
    .sort_index()
)

# Top ciudades
ventas_ciudad = (
    df_limpio
    .groupby("store_city")["net_sales"]
    .sum()
    .sort_values(ascending=False)
    .head(8)
)

# Categorías
ventas_categoria = (
    df_limpio
    .groupby("category")["net_sales"]
    .sum()
    .sort_values(ascending=False)
    .head(8)
)

# Tipo de cliente
ventas_cliente = (
    df_limpio
    .groupby("customer_type")["net_sales"]
    .sum()
    .sort_values(ascending=False)
)

# Método de pago
ventas_pago = (
    df_limpio
    .groupby("payment_mode")["net_sales"]
    .sum()
    .sort_values(ascending=False)
)

# ============================================================
# PCA
# ============================================================

pca_animacion = pca_df.copy()

# Tomamos una muestra para que la animación sea fluida
if len(pca_animacion) > 3000:
    pca_animacion = pca_animacion.sample(
        3000,
        random_state=42
    )

# ============================================================
# CREAR FIGURA
# ============================================================

fig = plt.figure(
    figsize=(18, 10)
)

fig.suptitle(
    "DASHBOARD ANIMADO - ANÁLISIS DE VENTAS",
    fontsize=20,
    fontweight="bold"
)

# ============================================================
# CREAR 9 RECUADROS
# ============================================================

ax1 = fig.add_subplot(3, 3, 1)
ax2 = fig.add_subplot(3, 3, 2)
ax3 = fig.add_subplot(3, 3, 3)
ax4 = fig.add_subplot(3, 3, 4)
ax5 = fig.add_subplot(3, 3, 5)
ax6 = fig.add_subplot(3, 3, 6)
ax7 = fig.add_subplot(3, 3, 7)
ax8 = fig.add_subplot(3, 3, 8)
ax9 = fig.add_subplot(3, 3, 9)

# ============================================================
# FUNCIÓN DE ACTUALIZACIÓN
# ============================================================

def actualizar(frame):

    # --------------------------------------------------------
    # LIMPIAR LOS GRÁFICOS
    # --------------------------------------------------------

    ax1.clear()
    ax2.clear()
    ax3.clear()
    ax4.clear()
    ax5.clear()
    ax6.clear()
    ax7.clear()
    ax8.clear()
    ax9.clear()

    # --------------------------------------------------------
    # 1. VENTAS MENSUALES
    # --------------------------------------------------------

    cantidad_meses = min(
        frame + 1,
        len(ventas_mes)
    )

    datos_ventas = ventas_mes.iloc[
        :cantidad_meses
    ]

    ax1.plot(
        datos_ventas.index,
        datos_ventas.values,
        marker="o",
        linewidth=2
    )

    ax1.set_title(
        "Ventas mensuales"
    )

    ax1.set_xlabel("Mes")
    ax1.set_ylabel("Ventas")

    ax1.grid(alpha=0.2)

    # --------------------------------------------------------
    # 2. GANANCIA MENSUAL
    # --------------------------------------------------------

    datos_ganancia = ganancia_mes.iloc[
        :cantidad_meses
    ]

    ax2.plot(
        datos_ganancia.index,
        datos_ganancia.values,
        marker="o",
        linewidth=2
    )

    ax2.set_title(
        "Ganancia mensual"
    )

    ax2.set_xlabel("Mes")
    ax2.set_ylabel("Ganancia")

    ax2.grid(alpha=0.2)

    # --------------------------------------------------------
    # 3. CIUDADES
    # --------------------------------------------------------

    cantidad_ciudades = min(
        frame + 1,
        len(ventas_ciudad)
    )

    datos_ciudades = ventas_ciudad.iloc[
        :cantidad_ciudades
    ]

    ax3.bar(
        datos_ciudades.index,
        datos_ciudades.values
    )

    ax3.set_title(
        "Top ciudades"
    )

    ax3.tick_params(
        axis="x",
        rotation=45,
        labelsize=8
    )

    # --------------------------------------------------------
    # 4. CATEGORÍAS
    # --------------------------------------------------------

    cantidad_categorias = min(
        frame + 1,
        len(ventas_categoria)
    )

    datos_categoria = ventas_categoria.iloc[
        :cantidad_categorias
    ]

    ax4.bar(
        datos_categoria.index,
        datos_categoria.values
    )

    ax4.set_title(
        "Ventas por categoría"
    )

    ax4.tick_params(
        axis="x",
        rotation=45,
        labelsize=8
    )

    # --------------------------------------------------------
    # 5. CLIENTES
    # --------------------------------------------------------

    cantidad_clientes = min(
        frame + 1,
        len(ventas_cliente)
    )

    datos_cliente = ventas_cliente.iloc[
        :cantidad_clientes
    ]

    ax5.bar(
        datos_cliente.index,
        datos_cliente.values
    )

    ax5.set_title(
        "Tipo de cliente"
    )

    ax5.tick_params(
        axis="x",
        rotation=45,
        labelsize=8
    )

    # --------------------------------------------------------
    # 6. MÉTODO DE PAGO
    # --------------------------------------------------------

    cantidad_pagos = min(
        frame + 1,
        len(ventas_pago)
    )

    datos_pago = ventas_pago.iloc[
        :cantidad_pagos
    ]

    ax6.bar(
        datos_pago.index,
        datos_pago.values
    )

    ax6.set_title(
        "Método de pago"
    )

    ax6.tick_params(
        axis="x",
        rotation=45,
        labelsize=8
    )

    # --------------------------------------------------------
    # 7. PCA ANIMADO
    # --------------------------------------------------------

    cantidad_puntos = int(
        len(pca_animacion)
        * ((frame + 1) / 30)
    )

    cantidad_puntos = min(
        cantidad_puntos,
        len(pca_animacion)
    )

    datos_pca = pca_animacion.iloc[
        :cantidad_puntos
    ]

    ax7.scatter(
        datos_pca["PC1"],
        datos_pca["PC2"],
        s=6,
        alpha=0.35
    )

    ax7.set_title(
        "PCA - Visualización"
    )

    ax7.set_xlabel("PC1")
    ax7.set_ylabel("PC2")

    ax7.grid(alpha=0.2)

    # --------------------------------------------------------
    # 8. K-MEANS ANIMADO
    # --------------------------------------------------------

    ax8.scatter(
        datos_pca["PC1"],
        datos_pca["PC2"],
        c=datos_pca["cluster"],
        s=6,
        alpha=0.4
    )

    ax8.set_title(
        "K-Means - Clusters"
    )

    ax8.set_xlabel("PC1")
    ax8.set_ylabel("PC2")

    ax8.grid(alpha=0.2)

    # --------------------------------------------------------
    # 9. GANANCIA
    # --------------------------------------------------------

    ganancias = (
        df_limpio["profit_amount"]
        .dropna()
    )

    # Se muestra progresivamente
    cantidad_hist = int(
        len(ganancias)
        * ((frame + 1) / 30)
    )

    cantidad_hist = max(
        cantidad_hist,
        1
    )

    cantidad_hist = min(
        cantidad_hist,
        len(ganancias)
    )

    datos_hist = ganancias.iloc[
        :cantidad_hist
    ]

    ax9.hist(
        datos_hist,
        bins=30
    )

    ax9.set_title(
        "Distribución de ganancias"
    )

    ax9.set_xlabel("Ganancia")
    ax9.set_ylabel("Frecuencia")

    ax9.grid(alpha=0.2)

    # --------------------------------------------------------
    # TÍTULO DINÁMICO
    # --------------------------------------------------------

    fig.suptitle(
        f"DASHBOARD ANIMADO - VENTA INDIA | FRAME {frame + 1}/30",
        fontsize=20,
        fontweight="bold"
    )

    plt.tight_layout(
        rect=[0, 0, 1, 0.94]
    )


# ============================================================
# CREAR ANIMACIÓN
# ============================================================

animacion_dashboard = FuncAnimation(
    fig,
    actualizar,
    frames=30,
    interval=500,
    repeat=True
)

# Evita que aparezca una figura estática adicional
plt.close(fig)

# ============================================================
# MOSTRAR ANIMACIÓN EN JUPYTER
# ============================================================

display(
    HTML(
        animacion_dashboard.to_jshtml()
    )
)
```
